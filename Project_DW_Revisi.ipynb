```python
import warnings

warnings.filterwarnings('ignore', category=FutureWarning)

output_path_indo = '/content/drive/My Drive/cleaned_covid_indo_data.csv'
output_path_usa = '/content/drive/My Drive/cleaned_covid_usa_data.csv'

df_indo = pd.read_csv(output_path_indo)
df_usa = pd.read_csv(output_path_usa)

df_indo['Date'] = pd.to_datetime(df_indo['Date'])
df_usa['data_as_of'] = pd.to_datetime(df_usa['data_as_of'])
df_usa['data_period_start'] = pd.to_datetime(df_usa['data_period_start'])
df_usa['data_period_end'] = pd.to_datetime(df_usa['data_period_end'])

df_usa = df_usa.loc[df_usa['data_period_end'] <= '2022-09-16']
df_usa = df_usa.rename(columns={'data_period_end': 'Date', 'COVID_deaths': 'New Deaths'})

df_agresi_indo = df_indo.groupby('Date').mean(numeric_only=True).reset_index()
df_agresi_usa = df_usa.groupby('Date').mean(numeric_only=True).reset_index()

df_indo_national = df_indo[df_indo['Location'] == 'Indonesia'].copy()
df_indo_national.sort_values(by='Date', inplace=True)

df_usa_national = df_usa[(df_usa['Jurisdiction_Residence'] == 'United States') & (df_usa['Group'] == 'total')].copy()
df_usa_national.sort_values(by='Date', inplace=True)

df_agresi_gabungan = df_gabungan.groupby('Date').mean(numeric_only=True).reset_index()
```

#DATA EXPLORATION AND VISUALISATION

```python
plt.figure(figsize=(15, 6))
sns.lineplot(x='Date', y='Total Cases', data=df_indo_national, label='Total Cases')
sns.lineplot(x='Date', y='Total Deaths', data=df_indo_national, label='Total Deaths')
plt.title('Total COVID-19 Cases and Deaths in Indonesia Over Time')
plt.xlabel('Date')
plt.ylabel('Count')
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15, 6))
sns.lineplot(x='Date', y='New Recovered', data=df_indo_national, label='New Recovered')
plt.title('New COVID-19 Recovered Cases in Indonesia Over Time')
plt.xlabel('Date')
plt.ylabel('Count')
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(18, 5))
plt.subplot(1, 3, 1)
sns.histplot(df_indo_national['New Cases'], bins=30, kde=True)
plt.title('Distribution of New Cases (Indonesia)')
plt.xlabel('New Cases')
plt.ylabel('Frequency')

plt.subplot(1, 3, 2)
sns.histplot(df_indo_national['New Deaths'], bins=30, kde=True)
plt.title('Distribution of New Deaths (Indonesia)')
plt.xlabel('New Deaths')
plt.ylabel('Frequency')

plt.subplot(1, 3, 3)
sns.histplot(df_indo_national['New Recovered'], bins=30, kde=True)
plt.title('Distribution of New Recovered (Indonesia)')
plt.xlabel('New Recovered')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
df_indo_provinces = df_indo[df_indo['Location'] != 'Indonesia'].copy()
top_indo_locations = df_indo_provinces.groupby('Location')['Total Cases'].max().nlargest(10).index

plt.figure(figsize=(15, 7))
sns.barplot(x='Total Cases', y='Location', data=df_indo_provinces[df_indo_provinces['Location'].isin(top_indo_locations)].sort_values(by='Total Cases', ascending=False), estimator=max, errorbar=None, palette='viridis')
plt.title('Top 10 Locations by Total COVID-19 Cases in Indonesia')
plt.xlabel('Total Cases')
plt.ylabel('Location')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15, 6))
sns.lineplot(x='Date', y='New Deaths_x', data=df_agresi_gabungan, label='New Deaths (USA from Merged)')
plt.title('Aggregated New COVID-19 Deaths in USA Over Time (from Merged Data)')
plt.xlabel('Date')
plt.ylabel('Number of Deaths')
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15, 6))
sns.lineplot(x='Date', y='crude_COVID_rate', data=df_agresi_gabungan, label='Crude COVID Rate (from Merged)')
sns.lineplot(x='Date', y='aa_COVID_rate', data=df_agresi_gabungan, label='AA COVID Rate (from Merged)')
plt.title('Aggregated COVID-19 Rates in USA Over Time (from Merged Data)')
plt.xlabel('Date')
plt.ylabel('Rate')
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df_usa_states = df_usa[~df_usa['Jurisdiction_Residence'].isin(['United States', 'Region 1', 'Region 2', 'Region 3', 'Region 4', 'Region'])].copy()

top_usa_jurisdictions = df_usa_states.groupby('Jurisdiction_Residence')['New Deaths'].sum().nlargest(10).index

plt.figure(figsize=(15, 7))
sns.barplot(x='New Deaths', y='Jurisdiction_Residence', data=df_usa_states[df_usa_states['Jurisdiction_Residence'].isin(top_usa_jurisdictions)].sort_values(by='New Deaths', ascending=False), estimator=sum, errorbar=None, palette='magma')
plt.title('Top 10 Jurisdictions by Total COVID-19 Deaths in USA')
plt.xlabel('Total New Deaths')
plt.ylabel('Jurisdiction of Residence')
plt.tight_layout()
plt.show()
```
